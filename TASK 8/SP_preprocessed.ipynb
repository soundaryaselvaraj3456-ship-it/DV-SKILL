```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/studentperformance_preprocessed.csv")
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
grouped = df.groupby(
    ['parental level of education', 'lunch']
)[['math score', 'reading score', 'writing score']].mean()

print(grouped)
```

```python
grouped.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Average Test Scores by Parental Education and Lunch Type')
plt.xlabel('Parental Education Level')
plt.ylabel('Average Score')
plt.xticks(rotation=45)
plt.legend(title='Subject')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.boxplot(
    x='test preparation course',
    y='math score',
    data=df
)

plt.title('Math Score Variation by Test Preparation')
plt.xlabel('Test Preparation Course')
plt.ylabel('Math Score')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    x='test preparation course',
    y='reading score',
    data=df
)

plt.title('Reading Score Variation by Test Preparation')
plt.xlabel('Test Preparation Course')
plt.ylabel('Reading Score')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    x='test preparation course',
    y='writing score',
    data=df
)

plt.title('Writing Score Variation by Test Preparation')
plt.xlabel('Test Preparation Course')
plt.ylabel('Writing Score')
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x='math score',
    y='reading score',
    data=df
)

plt.title('Math Score vs Reading Score')
plt.xlabel('Math Score')
plt.ylabel('Reading Score')
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x='math score',
    y='writing score',
    data=df
)

plt.title('Math Score vs Writing Score')
plt.xlabel('Math Score')
plt.ylabel('Writing Score')
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x='reading score',
    y='writing score',
    data=df
)

plt.title('Reading Score vs Writing Score')
plt.xlabel('Reading Score')
plt.ylabel('Writing Score')
plt.show()
```

```python
scores = df[
    ['math score', 'reading score', 'writing score']
]
```

```python
correlation = scores.corr()

print(correlation)

```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Student Scores')
plt.show()
```

```python
df['average score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
) / 3

print(
    df.groupby('test preparation course')['average score'].mean()
)
```

```python
print("Average scores by lunch type:")
print(
    df.groupby('lunch')[
        ['math score', 'reading score', 'writing score']
    ].mean()
)

print("\nAverage scores by parental education:")
print(
    df.groupby('parental level of education')[
        ['math score', 'reading score', 'writing score']
    ].mean()
)

print("\nAverage scores by test preparation:")
print(
    df.groupby('test preparation course')[
        ['math score', 'reading score', 'writing score']
    ].mean()
)
```
